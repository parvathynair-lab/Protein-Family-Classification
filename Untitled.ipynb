```python
import os,io,time,warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import requests
from scipy.cluster.hierarchy import linkage,dendrogram,fcluster,cophenet
from scipy.spatial.distance import pdist,cdist
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.metrics import (adjusted_rand_score,normalized_mutual_info_score,homogeneity_score, completeness_score, silhouette_score,
                             accuracy_score, confusion_matrix, classification_report)
warnings.filterwarnings("ignore")
plt.rcParams.update({"figure.dpi": 110, "axes.grid": False, "font.size": 10})

FAMILIES = {
    "PF00089": "Trypsin (serine protease)",
    "PF00001": "GPCR (rhodopsin-like)",
    "PF00072": "Response regulator receiver domain",
    "PF00018": "SH3 domain",
    "PF00076": "RNA recognition motif",
    "PF00096": "Zinc finger, C2H2 type"
}
N_PER_FAMILY = 150            
MIN_LEN, MAX_LEN = 50, 1500    
FEATURES   = "AAC+DPC"         
SCALE      = True              
            
LINKAGE    = "ward"            
TEST_SIZE  = 0.20          

K = 9     
print("Configuration loaded. Target clusters K =", K)
```

```python
DATA_DIR="data"
for d in (DATA_DIR,):
    os.makedirs(d, exist_ok=True)
def fetch_family(pfam_id, max_records=500):
    url = "https://rest.uniprot.org/uniprotkb/search"
    params = {
        "query": f"(xref:pfam-{pfam_id}) AND (reviewed:true)",
        "fields": "accession,id,protein_name,organism_name,length,sequence",
        "format": "tsv",
        "size": max_records,         
    }
    r = requests.get(url, params=params, timeout=60)
    r.raise_for_status()
    return pd.read_csv(io.StringIO(r.text), sep="\t")
CSV_PATH = os.path.join(DATA_DIR, "protein_sequences.csv")
if os.path.exists(CSV_PATH):
    df_raw = pd.read_csv(CSV_PATH)
    print("Loaded cached dataset:", CSV_PATH)
else:
    frames = []
    for pfam_id, name in FAMILIES.items():
        d = fetch_family(pfam_id)
        d["family"], d["pfam"] = name, pfam_id
        frames.append(d)
        print(f"{name:28s} ({pfam_id}): {len(d)} sequences downloaded")
        time.sleep(1)        
    df_raw = pd.concat(frames, ignore_index=True)
    df_raw.to_csv(CSV_PATH, index=False)
    print("Saved to", CSV_PATH)

print("\nRaw dataset shape:", df_raw.shape)
df_raw.head()
```

```python
SEED=42
AA = "ACDEFGHIKLMNPQRSTVWY"         
AA_SET = set(AA)
df = df_raw.copy()
print("start                       :", len(df))
df = df.drop_duplicates(subset="Entry", keep=False)
print("after removing ambiguous     :", len(df))
df = df.drop_duplicates(subset="Sequence")
print("after removing duplicates    :", len(df))
df = df[df["Sequence"].apply(lambda s: set(s) <= AA_SET)]
print("after removing odd residues  :", len(df))
df["Length"] = df["Sequence"].str.len()
df = df[df["Length"].between(MIN_LEN, MAX_LEN)]
print("after length filter          :", len(df))
parts = [g.sample(min(len(g), N_PER_FAMILY), random_state=SEED) for _, g in df.groupby("family")]
df = pd.concat(parts).reset_index(drop=True)
print("final (balanced) dataset     :", len(df))
print("\nSequences per family:")
print(df["family"].value_counts().to_string())
df.to_csv(os.path.join(DATA_DIR, "protein_sequences_clean.csv"), index=False)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
df["family"].value_counts().plot.barh(ax=ax[0], color="steelblue")
ax[0].set_title("Sequences per family"); ax[0].set_xlabel("count")
for fam, g in df.groupby("family"):
    ax[1].hist(g["Length"], bins=30, alpha=0.5, label=fam)
ax[1].set_title("Sequence length distribution"); ax[1].set_xlabel("length (amino acids)")
ax[1].legend(fontsize=7)
plt.tight_layout(); plt.show()
```

```python
AA_IDX = {a: i for i, a in enumerate(AA)}
def aac(seq):
    """20-dim vector: fraction of every amino acid in the sequence."""
    v = np.zeros(20)
    for ch in seq:
        v[AA_IDX[ch]] += 1
    return v / len(seq)
def dpc(seq):
    """400-dim vector: fraction of every ordered pair of adjacent amino acids."""
    m = np.zeros((20, 20))
    for a, b in zip(seq, seq[1:]):         
        m[AA_IDX[a], AA_IDX[b]] += 1
    return (m / (len(seq) - 1)).ravel()
X_aac = np.vstack([aac(s) for s in df["Sequence"]])
X_dpc = np.vstack([dpc(s) for s in df["Sequence"]])
# a quick sanity check on a tiny sequence

print("aac('AAC') -> A:", aac("AAC")[AA_IDX['A']], " C:", aac("AAC")[AA_IDX['C']], "(expected 0.667 and 0.333)")
print("AAC matrix:", X_aac.shape, "| DPC matrix:", X_dpc.shape)
def get_features(kind):
    return {"AAC": X_aac, "DPC": X_dpc, "AAC+DPC": np.hstack([X_aac, X_dpc])}[kind]
y_names = df["family"].values                    
classes = sorted(df["family"].unique())

# View the first 5 sequences and their features
print("AAC Sample Data:\n", X_aac[:5])
print("DPC Sample Data:\n", X_dpc[:5])
```

```python
idx = np.arange(len(df))
idx_tr, idx_te = train_test_split(idx, test_size=TEST_SIZE, stratify=y_names, random_state=SEED)
y_tr, y_te = y_names[idx_tr], y_names[idx_te]
print("train:", len(idx_tr), "| test:", len(idx_te))
```

```python
def embed(X_train, X_test, scale=True, n_pca=100):
    scaler = StandardScaler().fit(X_train) if scale else None
    A, B = (scaler.transform(X_train), scaler.transform(X_test)) if scale else (X_train, X_test)
    n_pca = min(n_pca, A.shape[1], A.shape[0] - 1)
    pca = PCA(n_components=n_pca, random_state=SEED).fit(A)
    return pca.transform(A), pca.transform(B), scaler, pca
X_all = get_features(FEATURES)
Z_tr, Z_te, scaler, pca = embed(X_all[idx_tr], X_all[idx_te], SCALE, 200)
print(f"{FEATURES}: {X_all.shape[1]} features  ->  {Z_tr.shape[1]} principal components")
print(f"Variance retained by these components: {pca.explained_variance_ratio_.sum():.1%}")
plt.figure(figsize=(6, 3.5))
plt.plot(np.cumsum(pca.explained_variance_ratio_), marker="o", ms=3)
plt.xlabel("number of components"); plt.ylabel("cumulative explained variance")
plt.title("PCA explained variance"); 
plt.show()
```

```python
# Compare linkage methods on the training set
D_tr = pdist(Z_tr)                       # all pairwise Euclidean distances (condensed form)

rows = []
for method in ["single", "complete", "average", "ward"]:
    Zm = linkage(Z_tr, method=method)    # Zm = linkage matrix: each row = one merge
    coph, _ = cophenet(Zm, D_tr)         # cophenetic correlation coefficient
    lab = fcluster(Zm, K, "maxclust")    # cut the tree so that exactly K clusters remain
    rows.append({"linkage": method,
                 "cophenetic corr.": coph,
                 "ARI": adjusted_rand_score(y_tr, lab),
                 "NMI": normalized_mutual_info_score(y_tr, lab),
                 "silhouette": silhouette_score(Z_tr, lab) if len(set(lab)) > 1 else np.nan})
link_cmp = pd.DataFrame(rows).round(3)
print(f"Chosen linkage for the rest of the notebook: '{LINKAGE}'")
link_cmp
```

```python
# Build the final linkage matrix with the chosen method
Z = linkage(Z_tr, method=LINKAGE)
print("Linkage matrix shape:", Z.shape, "(n-1 merges x 4 columns)")
print("Columns = [cluster A, cluster B, distance between them, #points in the new cluster]")
print(Z[:3].round(3), "\n...\n", Z[-3:].round(3))
```

```python
# Cut height that gives exactly K clusters = midway between the (K-1)th and Kth last merge
cut_height = (Z[-(K - 1), 2] + Z[-K, 2]) / 2
# (a) truncated dendrogram: only the last 30 merges, easier to read
fig, ax = plt.subplots(figsize=(12, 6))
dendrogram(Z, truncate_mode="lastp", p=30, show_leaf_counts=True,
           color_threshold=cut_height, above_threshold_color="grey", ax=ax)
ax.axhline(cut_height, ls="--", c="red", label=f"cut for k = {K}")
ax.set_title(f"Truncated dendrogram ({LINKAGE} linkage)  – numbers = proteins in the branch")
ax.set_xlabel("branch (size)"); ax.set_ylabel("merge distance"); ax.legend()
fig.tight_layout()
plt.tight_layout();plt.show()

# (b) full dendrogram + strip showing the true family of every leaf
fig = plt.figure(figsize=(13, 5.5))
gs = fig.add_gridspec(2, 1, height_ratios=[10, 0.7], hspace=0.03)
ax1, ax2 = fig.add_subplot(gs[0]), fig.add_subplot(gs[1])

den = dendrogram(Z, no_labels=True, color_threshold=cut_height, above_threshold_color="grey", ax=ax1)
ax1.set_ylabel("merge distance"); ax1.set_xticks([])
ax1.set_title("Full dendrogram – colour bar below = TRUE family of each protein")

leaf_order = den["leaves"]                              # order of proteins along the x-axis
family_code = np.array([classes.index(f) for f in y_tr])[leaf_order]
cmap = plt.get_cmap("tab10", len(classes))
ax2.imshow(family_code[None, :], aspect="auto", cmap=cmap, vmin=-0.5, vmax=len(classes) - 0.5,
           extent=(0, 10 * len(leaf_order), 0, 1), interpolation="nearest")
ax2.set_yticks([]); ax2.set_xticks([])
handles = [plt.Rectangle((0, 0), 1, 1, color=cmap(i)) for i in range(len(classes))]
ax2.legend(handles, classes, ncol=len(classes), fontsize=7, loc="upper center", bbox_to_anchor=(0.5, -0.2))
plt.show()
```

```python
def cluster_metrics(y_true, labels, Zspace=None):
    """Collect all external (label-based) and, optionally, the internal silhouette metric."""
    m = {"ARI": adjusted_rand_score(y_true, labels),
         "NMI": normalized_mutual_info_score(y_true, labels),
         "Homogeneity": homogeneity_score(y_true, labels),
         "Completeness": completeness_score(y_true, labels)}
    if Zspace is not None and len(set(labels)) > 1:
        m["Silhouette"] = silhouette_score(Zspace, labels)
    return m

labels_tr = fcluster(Z, K, "maxclust")                       
# Contingency table: rows = clusters, columns = true families
ct = pd.crosstab(pd.Series(labels_tr, name="cluster"), pd.Series(y_tr, name="true family"))
print(ct.to_string())

# Name each cluster after its majority family
cluster_to_family = ct.idxmax(axis=1).to_dict()
print("\nCluster -> majority family:")
for c, f in cluster_to_family.items():
    print(f"  cluster {c}: {f:28s} purity = {ct.loc[c].max() / ct.loc[c].sum():.2f}  (n = {ct.loc[c].sum()})")

pred_tr = np.array([cluster_to_family[c] for c in labels_tr])
print(f"\nTRAIN accuracy (cluster purity): {accuracy_score(y_tr, pred_tr):.3f}")

train_metrics = cluster_metrics(y_tr, labels_tr, Z_tr)
pd.Series(train_metrics).round(3).to_frame("TRAIN")
```

```python
def plot_heatmap(M, xlabels, ylabels, title, xlab, ylab, fname, cmap="Blues", fmt="d"):
    """Annotated heatmap with matplotlib only (no seaborn needed)."""
    fig, ax = plt.subplots(figsize=(1.1 * len(xlabels) + 2, 0.7 * len(ylabels) + 2))
    im = ax.imshow(M, cmap=cmap)
    ax.set_xticks(range(len(xlabels))); ax.set_xticklabels(xlabels, rotation=40, ha="right")
    ax.set_yticks(range(len(ylabels))); ax.set_yticklabels(ylabels)
    thresh = np.max(M) / 2
    for i in range(M.shape[0]):
        for j in range(M.shape[1]):
            ax.text(j, i, format(M[i, j], fmt), ha="center", va="center",
                    color="white" if M[i, j] > thresh else "black", fontsize=9)
    ax.set_title(title); ax.set_xlabel(xlab); ax.set_ylabel(ylab)
    plt.colorbar(im, fraction=0.046); plt.tight_layout()
    plt.show()

plot_heatmap(ct.values, list(ct.columns), [f"cluster {c}" for c in ct.index],
             "Training set: clusters vs true families", "true family", "cluster",
             "07_contingency_train")
```

```python
def scatter_pair(P, true_lab, clus_lab, title, fname):
    fig, ax = plt.subplots(1, 2, figsize=(13, 5))
    for i, c in enumerate(classes):
        m = true_lab == c
        ax[0].scatter(P[m, 0], P[m, 1], s=14, alpha=0.8, label=c)
    ax[0].set_title(f"{title} – TRUE families"); ax[0].legend(fontsize=7)
    sc = ax[1].scatter(P[:, 0], P[:, 1], c=clus_lab, cmap="tab10", s=14, alpha=0.8)
    ax[1].set_title(f"{title} – hierarchical clusters (k={K})")
    handles, labels = sc.legend_elements()
    ax[1].legend(handles, labels, fontsize=7)
    for a in ax: a.set_xticks([]); a.set_yticks([])
    plt.tight_layout(); plt.show()

scatter_pair(Z_tr[:, :2], y_tr, labels_tr, "PCA", "08_pca_scatter")
```

```python
cluster_ids = np.array(sorted(cluster_to_family))
centroids = np.vstack([Z_tr[labels_tr == c].mean(axis=0) for c in cluster_ids])   # one centre per cluster

def assign_to_clusters(Znew):
    """Nearest-centroid assignment (Euclidean distance in PCA space)."""
    return cluster_ids[cdist(Znew, centroids).argmin(axis=1)]

labels_te = assign_to_clusters(Z_te)
pred_te = np.array([cluster_to_family[c] for c in labels_te])

test_acc = accuracy_score(y_te, pred_te)
print(f"TEST accuracy on unseen proteins: {test_acc:.3f}\n")
print(classification_report(y_te, pred_te, zero_division=0))

test_metrics = cluster_metrics(y_te, labels_te)           # silhouette skipped: centroid-assigned
comparison = pd.DataFrame({"TRAIN": train_metrics, "TEST": test_metrics}).round(3)
comparison.loc["Accuracy"] = [round(accuracy_score(y_tr, pred_tr), 3), round(test_acc, 3)]
comparison
```

```python
cm = confusion_matrix(y_te, pred_te, labels=classes)
plot_heatmap(cm, classes, classes, "Test set confusion matrix (unseen proteins)",
             "predicted family", "true family", "10_confusion_matrix_test")

# Per-family accuracy (recall) - which families are easy / hard to separate?
per_family = pd.Series(cm.diagonal() / cm.sum(axis=1), index=classes, name="recall").round(3)

per_family.to_frame()
```
